```python
from pyspark.sql.functions import *
```

```python
def age_band(dob_col):
    age = datediff(current_date(), dob_col) / 365.25
    return (
        when(age < 18, "0-17")
        .when(age < 40, "18-39")
        .when(age < 65, "40-64")
        .otherwise("65+")
    )
```

```python
def cost_flag(amount_col):
    return when(amount_col >= 50, "high").otherwise("normal")
```
